> ### <b>문제 1</b>. 설문조사를 통해 리커트 5점 척도 문항을 통해 아래와 같은 응답을 받았다. 가령 ``1=매우 그렇다, 2=다소 그렇다, 3=중립적, 4=다소 그렇지 않다, 5=매우 그렇지 않다.''라고 한다면, 이를 역부호화(reverse coding)하는 코드를 작성하시오.
```python
answers = [1, 3, 2, 4, 1, 5, 2, 3, 1, 4, 2]

````python
answers = [1, 3, 2, 4, 1, 5, 2, 3, 1, 4, 2]

#길게
recoded = []
for answer in answers:
    recoded.append(max(answers) + 1 - answer)
print(recoded)

#짧게
[max(answers) + 1 - answer for answer in answers]
````

> ### <b>문제 2</b>. 다음은 응답자의 직업 목록이다. 외부 모듈을 쓰지 말고, 딕셔너리(dict)로 직업별 빈도를 센 뒤 가장 많이 나온 직업과 그 빈도를 출력하는 코드를 작성하시오. 그리고 빈도가 가장 높은 직업이 둘 이상일 때 여러분의 코드가 무엇을 출력하는지, 그것이 분석 결과를 보고할 때 어떤 문제를 낳을 수 있는지 서술하시오.
>

```python
jobs = ["자영업", "학생", "학생", "회사원", "자영업", "학생", "무직", "회사원", "학생", "학생", "회사원", "무직", "회사원", "학생"]

````python
jobs = ["자영업", "학생", "학생", "회사원", "자영업", "학생", "무직", "회사원", "학생", "학생", "회사원", "무직", "회사원", "학생"]

freq = {}
for j in jobs:
    if j in freq:
        freq[j] = freq[j] + 1
    else:
        freq[j] = 1

max_job = None
max_count = 0
for job in freq:
    if freq[job] > max_count:
        max_job = job
        max_count = freq[job]

print(freq)
print(max_job, max_count)
````

> ### <b>문제 3</b>. 인구 89,000명인 충북의 한 도시에서 어느날 한 사람이 헛소문(rumor)을 지어냈다. 이제 하루가 지날 때마다 헛소문을 아는 사람 각각이 모르는 사람 2명에게 헛소문을 퍼트린다고 가정하자.
>
> (1) 소문을 아는 사람이 인구 전체에 이르는 데 며칠이 걸리는지 출력하는 코드를 작성하시오.\
> (2) 10일이 경과한 날에 몇 명이나 그 헛소문을 아는지 계산하는 코드를 작성하시오.\
> (3) 이 모형은 사회학적으로 어떤 함의를 갖는지를 분석적으로 간략히 논하시오.

````python
population = 89000
knowers = 1
for day in range(1, population + 1):
    knowers = knowers * 3
    if day == 10:
        print("10일째:", int(knowers))
    if knowers >= population:
        break
print(str(day)+"일이 걸림")
````
